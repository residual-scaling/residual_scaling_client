```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
from prompt_manager import PromptManager
```

```python
model_path = "/root/autodl-tmp/Meta-Llama-3-8B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForCausalLM.from_pretrained(
            model_path,
            device_map="auto",
            torch_dtype="auto"
        )
pipe = pipeline("text-generation", 
                    model=model, 
                    tokenizer=tokenizer,
                    do_sample=True,
                    temperature=0.7
                    )
pm = PromptManager()
```

```python
query = "Compounds that are capable of accepting electrons, such as o 2 or f2, are called what?"
options = ['residues', 'antioxidants', 'Oxygen', 'oxidants']
context = ["oxidants is NOT the answer"]
prompt = pm.generate_llama_percise_question_prompt(query, options, context)
print(prompt)
```

```python
result = pipe(prompt)
print(result)
```
