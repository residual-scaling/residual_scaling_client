```python
from experience_builder import ExperienceBuilder
from tester import Tester
from util import log_flush
from model.model_loader import load_model
import pandas as pd
from experiment_analyzer.single_run_analyzer import SingleRunAnalyzer
```

```python
model_path = "/root/autodl-tmp/Meta-Llama-3-8B-Instruct"
model = load_model("local", model_path)
```

```python
# hyper parameter
subject = "biology"
dataset="MMLU"
subset = "high_school_biology"
```

```python
builder = ExperienceBuilder(
    id="01",
    subject=subject,
    dataset=dataset,
    subset=subset,
    train_percent=0.8,
    model=model,
    trace_scores=True,
    save_distill_experience=True,
    save_false_experience=True,
)
```

```python
print(len(builder.train_set))
print(len(builder.test_set))
print(builder.dataloader.size)
```

```python
builder.start()
```

```python
builder.save_experience_db()
```

```python
analyzer = SingleRunAnalyzer(70)
```

```python
fio = open("./log/round20test.log", 'a')
for i in range(20):
    log_flush(fio, f"+++ round {i}")

    plain_tester = Tester(
        id=f"01-base_{i}",
        subject=subject,
        dataset=dataset,
        subset=subset,
        train_percent=0.8,
        model=model,
        analyzer=analyzer,
        trace_scores=True,
        use_experience=False,
    )
    experience_tester = Tester(
        id=f"01-experience_{i}",
        subject=subject,
        dataset=dataset,
        subset=subset,
        train_percent=0.8,
        model=model,
        analyzer=analyzer,
        trace_scores=True,
        use_experience=True,
    )
    
    p_accurate_count, _ = plain_tester.start()
    log_flush(fio, f"p_accurate_count {p_accurate_count}")
    
    e_accurate_count, _ = experience_tester.start()
    log_flush(fio, f"e_accurate_count {e_accurate_count}")
```
