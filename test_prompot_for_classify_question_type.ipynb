```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
```

```python
model_path = "/root/autodl-tmp/Meta-Llama-3-8B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForCausalLM.from_pretrained(
            model_path,
            device_map="auto",
            torch_dtype="auto"
        )
pipe = pipeline("text-generation", 
                    model=model, 
                    tokenizer=tokenizer,
                    do_sample=True,
                    temperature=0.7
                    )
```

```python
question = "Which major river flows through 10 countries in Central and Eastern Europe, forms part of the border between Slovakia and Hungary, and empties into the Black Sea? Name the river and two of the four capital cities it passes through."
prompt = f"""<|begin_of_text|><|start_header_id|>system<|end_header_id|>
Classify this question into exactly one category: "geography", "biology", "economy", or "others". Respond with only the category name as one word.<|eot_id|>
<|start_header_id|>user<|end_header_id|>
Question: {question}<|eot_id|>
<|start_header_id|>assistant<|end_header_id|>"""
```

```python
pipe(prompt)
```
